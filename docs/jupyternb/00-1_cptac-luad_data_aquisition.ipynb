```r
clin <- fread("./data/cptac-luad/S046_Lung_Adenocarcinoma.csv")
clin
```

## Extract tumor samples

```r
channel_levels <- c(
  "126", "127N", "127C", "128N", "128C",
  "129N", "129C", "130N", "130C", "131"
)
```

```r
phospho_ann <- unique(
  clin[
    Batch == "Phosphoproteome",
    .(
      Mixture,
      Channel,
      Subject,
      Condition,
      BioReplicate,
      Cohort
    )
  ]
)

phospho_ann[, channel_order := match(Channel, channel_levels)]
setorder(phospho_ann, Mixture, channel_order)

phospho_ann <- phospho_ann[Condition != "Norm"]
phospho_ann[, run_id := unique(phospho$run_id)]
phospho_ann[, subject_id := sub("^Patient_", "", Subject)]
phospho_ann[, subject_id := gsub("-", ".", subject_id)]
```

```r
phospho_ann <- phospho_ann[Condition == "Lung Cancer"]
phospho <- phospho[run_id %in% phospho_ann$run_id]
uniqueN(phospho$run_id)
```

```r
proteo_ann <- unique(
  clin[
    Batch == "Proteome",
    .(
      Mixture,
      Channel,
      Subject,
      Condition,
      BioReplicate,
      Cohort
    )
  ]
)

proteo_ann[, channel_order := match(Channel, channel_levels)]
setorder(proteo_ann, Mixture, channel_order)

proteo_ann <- proteo_ann[Condition != "Norm"]
proteo_ann[, run_id := unique(proteo$run_id)]
proteo_ann[, subject_id := sub("^Patient_", "", Subject)]
proteo_ann[, subject_id := gsub("-", ".", subject_id)]
```

```r
proteo_ann <- proteo_ann[Condition == "Lung Cancer"]
proteo <- proteo[run_id %in% proteo_ann$run_id]
uniqueN(proteo$run_id)
```

```r
common_subjects <- intersect(
  phospho_ann$subject_id,
  proteo_ann$subject_id
)

length(common_subjects)
```

## Exclude patients with multiple runs

```r
dup_subjects_phospho <- phospho_ann[
  subject_id %in% common_subjects,
  .N,
  by = subject_id
][
  N > 1,
  subject_id
]

dup_subjects_proteo <- proteo_ann[
  subject_id %in% common_subjects,
  .N,
  by = subject_id
][
  N > 1,
  subject_id
]

dup_subjects <- union(
  dup_subjects_phospho,
  dup_subjects_proteo
)

dup_subjects
```

```r
common_subjects <- setdiff(
  common_subjects,
  dup_subjects
)

length(common_subjects)
```

```r
phospho <- phospho[
    run_id %in% phospho_ann[
        subject_id %in% common_subjects,
        run_id
        ]
    ]
uniqueN(phospho$run_id)
```

```r
proteo <- proteo[
    run_id %in% proteo_ann[
        subject_id %in% common_subjects,
        run_id
        ]
    ]
uniqueN(proteo$run_id)
```

```r
saveRDS(phospho, "./data/cptac-luad/CPTAC_LUAD_phospho.rds")
saveRDS(proteo,  "./data/cptac-luad/CPTAC_LUAD_proteo.rds")
```
