```r
de <- read_xlsx(xlsx, sheet="Supplemental_table_S5_G12Ci_Pho")
de
```

```r
meta_proc <- meta %>%
  mutate(
    Position = as.integer(Position),
    
    gene_from_phosname = sub(" [STY][0-9]+$", "", phos.name),
    site_from_phosname = sub("^.* ", "", phos.name),

    phosphosite = site_from_phosname,

    gene_from_fasta = sub(
      ".* GN=([^ ]+).*",
      "\\1",
      Fasta_headers
    ),

    gene_id = ifelse(
      grepl(" GN=", Fasta_headers),
      gene_from_fasta,
      gene_from_phosname
    ),

    protein_id = Uniprot_ID,
      
    site_id = paste(gene_id, protein_id, phosphosite, sep = ":"),

    peptide_id = site_id,
    modified_peptide_sequence = site_id,
    peptide_sequence = site_id,
  ) %>%
  select(
    id,
    phos.name,
    gene_id,
    protein_id,
    phosphosite,
    site_id,
    peptide_id,
    modified_peptide_sequence,
    peptide_sequence,
    Localization_prob,
    PEP,
    Reverse,
    Potential_contaminant
  )

de_int <- de %>%
  left_join(
    meta_proc,
    by = c("id", "phos.name")
  )
```

```r
mvespa_long_t <- de_int %>%
  filter(!is.na(site_id)) %>%
  filter(!is.na(t)) %>%
  filter(is.na(Reverse) | Reverse != "+") %>%
  filter(is.na(Potential_contaminant) | Potential_contaminant != "+") %>%
  mutate(
    condition = paste(cell.line, Comparison, time, sep = "__")
  ) %>%
  select(site_id, condition, t) %>%
  group_by(site_id, condition) %>%
  summarise(
    value = mean(t, na.rm = TRUE),
    .groups = "drop"
  )

mvespa_mat_t <- mvespa_long_t %>%
  pivot_wider(
    names_from = condition,
    values_from = value
  ) %>%
  as.data.frame()

rownames(mvespa_mat_t) <- mvespa_mat_t$site_id
mvespa_mat_t$site_id <- NULL

mvespa_mat_t <- as.matrix(mvespa_mat_t)
storage.mode(mvespa_mat_t) <- "numeric"

dim(mvespa_mat_t)
mvespa_mat_t[1:5, 1:min(5, ncol(mvespa_mat_t))]
```
